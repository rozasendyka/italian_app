```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
file_path = r"C:\Users\roza0\Desktop\projects\italian_app\data\sentences\ita.txt"

data = pd.read_csv(file_path, sep="\t")
data.columns = ["english", "italian", "attrib"]
data.drop(["attrib"], inplace=True, axis=1)

print(data.head())
```

```python
for col in ["english", "italian"]:
    # clean the sentences: lowercase and remove punctuation
    data[f"{col}_clean"] = (
        data[col]
        
        .str.lower()
        .str.replace(r"[^\w\s]", "", regex=True)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )
    # count n words
    data[f"{col}_n_words"] = (
        data[f"{col}_clean"]
        .str.split()
        .str.len()
    )

# classify difficulty level
def classify_length(n):
    if n <= 5:
        return "easy"
    elif n <= 10:
        return "intermediate"
    elif n <= 25:
        return "advanced"
    else:
        return "extreme"
        
data["level"] = data["english_n_words"].apply(classify_length)
   
# Id column for distinct English phrases
data["id"] = pd.factorize(data["english"])[0]
```

```python
# Subset the data to keep one entry per English phrase
data_clean = data.drop_duplicates(subset="id", keep="first").copy()
```

```python
# Save data clean
file_clean_path = r"C:\Users\roza0\Desktop\projects\italian_app\data\sentences\ita_clean.csv"
data_clean.to_csv(file_clean_path)
```
