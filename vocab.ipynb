```python
# Get the list of all files and directories
path = r"C:\Users\roza0\Desktop\projects\italian_app\data\vocab"
files = os.listdir(path)
print(files)
```

```python
data = pd.DataFrame()
for f in files:
    dat =  pd.read_csv(os.path.join(path, f), sep="\t")
    dat = pd.concat(
                    [dat, dat.assign(cat1="all")],
                    ignore_index=True
                )
    print(dat.head())
    data = pd.concat([data, dat], ignore_index = True)
```

```python
if "current_word" not in st.session_state:
    st.session_state.current_word = data.sample(1).iloc[0]

word = st.session_state.current_word

st.header(word["english"])

answer = st.text_input("Italian")

if st.button("Check"):

    if answer.lower().strip() == word["italian"].lower():
        st.success("Correct!")
    else:
        st.error(f"Correct answer: {word['italian']}")
```
